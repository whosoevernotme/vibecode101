````python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_auc_score

# 1. Copy data to avoid overwriting original df
data = df.copy()

# 2. Replace impossible zeros with NaN for specified columns
impossible_zero_cols = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
data[impossible_zero_cols] = data[impossible_zero_cols].replace(0, np.nan)

# 3. Impute missing values with Median
imputer = SimpleImputer(strategy="median")
data[impossible_zero_cols] = imputer.fit_transform(data[impossible_zero_cols])

# 4. Split into features and target
X_ai = data.drop(columns=["Outcome"])
y_ai = data["Outcome"]
X_train, X_test, y_train, y_test = train_test_split(X_ai, y_ai, test_size=0.2, random_state=42)

# 5. Train Random Forest Classifier
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# 6. Evaluate and report performance suitable for medical screening
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:, 1]

tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()
sensitivity = tp / (tp + fn)
specificity = tn / (tn + fp)

print("=== Performance Metrics ===")
print(f"Accuracy: {accuracy_score(y_test, y_pred):.3f}")
print(f"ROC AUC: {roc_auc_score(y_test, y_prob):.3f}")
print(f"Sensitivity (Recall for Diabetes): {sensitivity:.3f}")
print(f"Specificity: {specificity:.3f}")
print("\n--- Confusion Matrix ---")
print(f"True Negative (TN): {tn}  |  False Positive (FP): {fp}")
print(f"False Negative (FN): {fn}  |  True Positive (TP): {tp}")
print("\n--- Classification Report ---")
print(classification_report(y_test, y_pred))
````

````python
# @title ✅ เฉลย — หากไม่สามารถทำแบบฝึกหัดได้ ให้รันเซลล์นี้ (ดับเบิลคลิกเพื่อแสดงโค้ด) { display-mode: "form" }
# ตัวอย่างคำตอบที่ผู้ช่วยปัญญาประดิษฐ์มักให้สำหรับคำสั่งในขั้นที่ 1 (รันได้แต่ไม่สมบูรณ์ โดยเจตนา
# เพื่อใช้ตรวจสอบในขั้นที่ 2 ในกรณีที่โค้ดของผู้เรียนรันไม่สำเร็จ)
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

data = df.copy()

# Replace impossible zeros with NaN
cols_with_zeros = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
data[cols_with_zeros] = data[cols_with_zeros].replace(0, np.nan)

# Impute missing values with the median
data[cols_with_zeros] = data[cols_with_zeros].fillna(data[cols_with_zeros].median())

# Split into features and target
X_ai = data.drop(columns=["Outcome"])
y_ai = data["Outcome"]
X_train, X_test, y_train, y_test = train_test_split(X_ai, y_ai, test_size=0.2)

# Train a random forest
model = RandomForestClassifier(n_estimators=100)
model.fit(X_train, y_train)

# Evaluate
y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nConfusion matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification report:\n", classification_report(y_test, y_pred))
````

## ขั้นที่ 3 — แก้ไข: เปรียบเทียบกับไปป์ไลน์อ้างอิง

ด้านล่างเป็นการเขียนไปป์ไลน์เดียวกันโดย **ไม่มีการรั่วไหลของข้อมูล** หลักสำคัญคือ `Pipeline` ของ scikit-learn ซึ่งรวมขั้นตอน "เติมค่าที่ขาดหาย → ฝึกโมเดล" ไว้ในออบเจกต์เดียว ค่ามัธยฐาน (median) ที่ใช้เติมจึงคำนวณจาก **ผู้ป่วยในชุดฝึกเท่านั้น** ทุกครั้ง รวมถึงภายในการตรวจสอบไขว้

*การตรวจสอบไขว้ (cross-validation) แบบ 5-fold* คือการแบ่งข้อมูลเป็น 5 ส่วน ใช้ 4 ส่วนในการฝึกและอีก 1 ส่วนในการทดสอบ ทำซ้ำ 5 รอบแล้วหาค่าเฉลี่ย ทำให้ได้คะแนนที่มีเสถียรภาพมากกว่าการแบ่งข้อมูลครั้งเดียว

**⭐ แบบฝึกหัดที่ 2 (Exercise 2) — เติมช่องว่าง 5 ช่อง** (แต่ละช่องเป็นคำเดียวหรือสูตรสั้น คำใบ้ระบุสิ่งที่ต้องพิมพ์ไว้แล้ว) หากไม่สามารถทำได้ ให้รันเซลล์ ✅ เฉลย ซึ่งเป็นไปป์ไลน์อ้างอิงฉบับสมบูรณ์

````python
# ⭐ Exercise 2 — แบบฝึกหัดที่ 2: ไปป์ไลน์อ้างอิงที่ไม่มี data leakage — เติมช่องว่าง 5 ช่องที่เป็น ...
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score, train_test_split
from sklearn.metrics import confusion_matrix

ZERO_MEANS_MISSING = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
X = df.drop(columns=["Outcome"]).replace({c: {0: np.nan} for c in ZERO_MEANS_MISSING})
y = df["Outcome"]

pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),                   # ช่องที่ 1: พิมพ์ "median" (รวมเครื่องหมายอัญประกาศ)
    ("model", RandomForestClassifier(n_estimators=300, random_state=42)),   # ช่องที่ 2: พิมพ์ 42
])

# ROC AUC จากการตรวจสอบไขว้แบบ 5-fold (imputer ถูก fit ใหม่ในทุก fold จึงไม่มี data leakage)
auc = cross_val_score(pipe, X, y, cv=5, scoring="roc_auc")          # ช่องที่ 3: พิมพ์ "roc_auc" (รวมเครื่องหมายอัญประกาศ)
print(f"Cross-validated ROC AUC: {auc.mean():.3f} ± {auc.std():.3f}")

# แบ่ง train/test แบบ stratified หนึ่งครั้ง เพื่อคำนวณ sensitivity และ specificity ที่เกณฑ์ตัดสินมาตรฐาน (0.5)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
pipe.fit(X_tr, y_tr)                                          # ค่า median คำนวณจากผู้ป่วยในชุดฝึกเท่านั้น
tn, fp, fn, tp = confusion_matrix(y_te, pipe.predict(X_te)).ravel()
sensitivity =tp / (tp + fn)            # ช่องที่ 4: ผู้ป่วยโรคเบาหวานที่ตรวจพบ / ผู้ป่วยโรคเบาหวานทั้งหมด -> พิมพ์  tp / (tp + fn)
specificity =tn / (tn + fp)            # ช่องที่ 5: ผู้ไม่เป็นโรคที่จำแนกถูกต้อง / ผู้ไม่เป็นโรคทั้งหมด -> พิมพ์  tn / (tn + fp)
print(f"ชุดทดสอบ: TP={tp} FN={fn} TN={tn} FP={fp}")
print(f"Sensitivity: {sensitivity:.3f}, Specificity: {specificity:.3f}")
````

````python
# @title ✅ เฉลย — หากไม่สามารถทำแบบฝึกหัดได้ ให้รันเซลล์นี้ (ดับเบิลคลิกเพื่อแสดงโค้ด) { display-mode: "form" }
# ไปป์ไลน์อ้างอิง (ไม่มี data leakage, ทำซ้ำได้, มีตัวชี้วัดสำหรับการตรวจคัดกรอง)
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score, train_test_split
from sklearn.metrics import confusion_matrix

ZERO_MEANS_MISSING = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
X = df.drop(columns=["Outcome"]).replace({c: {0: np.nan} for c in ZERO_MEANS_MISSING})
y = df["Outcome"]

pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", RandomForestClassifier(n_estimators=300, random_state=42)),
])

auc = cross_val_score(pipe, X, y, cv=5, scoring="roc_auc")
print(f"Cross-validated ROC AUC: {auc.mean():.3f} ± {auc.std():.3f}")

X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
pipe.fit(X_tr, y_tr)
tn, fp, fn, tp = confusion_matrix(y_te, pipe.predict(X_te)).ravel()
sensitivity = tp / (tp + fn)
specificity = tn / (tn + fp)
print(f"ชุดทดสอบ: TP={tp} FN={fn} TN={tn} FP={fp}")
print(f"Sensitivity: {sensitivity:.3f}, Specificity: {specificity:.3f}")
````

💡 **คำอธิบายผลลัพธ์** เมื่อเปรียบเทียบกับรายการตรวจสอบ ไปป์ไลน์อ้างอิงแก้ไขข้อ E (เติมค่าภายใน `Pipeline`) ข้อ F/G (กำหนด `random_state=42` ทุกตำแหน่ง ผลจึงเหมือนเดิมเมื่อรันซ้ำ) ข้อ H (การตรวจสอบไขว้แบบ 5-fold) และข้อ A–C (รายงาน ROC AUC, sensitivity และ specificity) ให้สังเกตค่า sensitivity ที่เกณฑ์ตัดสินมาตรฐาน 0.5 ซึ่งแสดงว่าโมเดลยัง **พลาดผู้ป่วยโรคเบาหวานในสัดส่วนสูง** (FN) ซึ่งไม่เหมาะสมสำหรับการตรวจคัดกรอง ขั้นที่ 4 จะแก้ไขประเด็นนี้

> สำหรับชุดข้อมูลนี้ การรั่วไหลของข้อมูลจากการเติมค่าก่อนแบ่งข้อมูลทำให้คะแนนเปลี่ยนเพียงเล็กน้อย แต่ในข้อมูลอื่น (เช่น การปรับสเกล การคัดเลือกตัวแปร (feature selection) หรือ oversampling ก่อนแบ่งข้อมูล หรือผู้ป่วยรายเดียวกันที่มารับบริการหลายครั้งปรากฏทั้งในชุดฝึกและชุดทดสอบ) การรั่วไหลของข้อมูลอาจทำให้ผลสูงเกินจริงอย่างมาก จึงควรปฏิบัติอย่างถูกต้องจนเป็นนิสัย

## ขั้นที่ 4 — ปรับปรุง: การเลือกเกณฑ์ตัดสินสำหรับการตรวจคัดกรอง

โมเดลให้ค่า *ความน่าจะเป็น (probability)* ของการเป็นโรคเบาหวานสำหรับผู้ป่วยแต่ละราย โดยค่าเริ่มต้น ผู้ที่มีค่าเกิน **0.5** จะถูกจำแนกว่ามีความเสี่ยง สำหรับการตรวจคัดกรอง โดยทั่วไปจะยอมรับผลบวกลวงที่เพิ่มขึ้นเพื่อลดการพลาดผู้ป่วย จึงใช้ **เกณฑ์ตัดสิน (threshold) ที่ต่ำลง**

เซลล์ด้านล่างแสดงค่า sensitivity และ specificity ที่เกณฑ์ตัดสินหลายค่า ซึ่งประเมิน **จากผู้ป่วยในชุดฝึกเท่านั้น** ด้วยการตรวจสอบไขว้ (การเลือกเกณฑ์จากชุดทดสอบถือเป็นการรั่วไหลของข้อมูลอีกรูปแบบหนึ่ง)

**⭐ แบบฝึกหัดที่ 3 (Exercise 3) — แก้ไขตัวเลขเพียงค่าเดียว:** กำหนด `THRESHOLD` เป็นค่าจากตารางที่ให้ **sensitivity ≥ 0.80** และมี specificity สูงที่สุด จากนั้นเซลล์จะประเมินค่าดังกล่าวกับผู้ป่วยในชุดทดสอบหนึ่งครั้ง

*(สำหรับผู้เรียนที่เขียนโปรแกรมได้: สามารถส่งคำสั่งถึงผู้ช่วยปัญญาประดิษฐ์แทน เช่น "Add threshold tuning with cross_val_predict on the training set so sensitivity is at least 0.80, then report test sensitivity and specificity" แล้วตรวจสอบคำตอบด้วยรายการตรวจสอบ)*

````python
# ⭐ Exercise 3 — แบบฝึกหัดที่ 3: เลือกเกณฑ์ตัดสินสำหรับการตรวจคัดกรอง (แก้ไขตัวเลขเพียงค่าเดียว)
from sklearn.model_selection import cross_val_predict

# ความน่าจะเป็นแบบ out-of-fold ของผู้ป่วยในชุดฝึกเท่านั้น (ไม่ใช้ข้อมูลชุดทดสอบ)
p_train_oof = cross_val_predict(pipe, X_tr, y_tr, cv=5, method="predict_proba")[:, 1]
table = pd.DataFrame([
    {"threshold": t,
     "sensitivity": ((p_train_oof >= t) & (y_tr == 1)).sum() / (y_tr == 1).sum(),
     "specificity": ((p_train_oof <  t) & (y_tr == 0)).sum() / (y_tr == 0).sum()}
    for t in [0.2, 0.25, 0.3, 0.35, 0.4, 0.45, 0.5]]).round(3)
print(table.to_string(index=False))

THRESHOLD = 0.3     # TODO: ระบุเกณฑ์จากตารางที่ sensitivity >= 0.80 และ specificity สูงที่สุด (คำใบ้: ประมาณ 0.3)

# ประเมินเกณฑ์ที่เลือกกับผู้ป่วยในชุดทดสอบที่แยกไว้ หนึ่งครั้ง
p_test = pipe.predict_proba(X_te)[:, 1]
flag = p_test >= THRESHOLD
test_sens = (flag & (y_te == 1)).sum() / (y_te == 1).sum()
test_spec = (~flag & (y_te == 0)).sum() / (y_te == 0).sum()
print(f"\nเกณฑ์ {THRESHOLD}: sensitivity ของชุดทดสอบ {test_sens:.3f}, specificity ของชุดทดสอบ {test_spec:.3f}")
````

````python
# @title ✅ เฉลย — หากไม่สามารถทำแบบฝึกหัดได้ ให้รันเซลล์นี้ (ดับเบิลคลิกเพื่อแสดงโค้ด) { display-mode: "form" }
# เลือกเกณฑ์โดยอัตโนมัติ: เกณฑ์สูงสุดที่ sensitivity (out-of-fold ของชุดฝึก) >= 0.80
from sklearn.model_selection import cross_val_predict

p_train_oof = cross_val_predict(pipe, X_tr, y_tr, cv=5, method="predict_proba")[:, 1]
table = pd.DataFrame([
    {"threshold": t,
     "sensitivity": ((p_train_oof >= t) & (y_tr == 1)).sum() / (y_tr == 1).sum(),
     "specificity": ((p_train_oof <  t) & (y_tr == 0)).sum() / (y_tr == 0).sum()}
    for t in [0.2, 0.25, 0.3, 0.35, 0.4, 0.45, 0.5]]).round(3)
print(table.to_string(index=False))

ok = table[table["sensitivity"] >= 0.80]
THRESHOLD = float(ok["threshold"].max()) if len(ok) else 0.2   # เกณฑ์สูงสุด = specificity สูงที่สุด

p_test = pipe.predict_proba(X_te)[:, 1]
flag = p_test >= THRESHOLD
test_sens = (flag & (y_te == 1)).sum() / (y_te == 1).sum()
test_spec = (~flag & (y_te == 0)).sum() / (y_te == 0).sum()
print(f"\nเกณฑ์ {THRESHOLD}: sensitivity ของชุดทดสอบ {test_sens:.3f}, specificity ของชุดทดสอบ {test_spec:.3f}")
````

💡 **คำอธิบายผลลัพธ์** การลดเกณฑ์ตัดสินทำให้ตรวจพบผู้ป่วยโรคเบาหวานได้มากขึ้น (sensitivity เพิ่มขึ้น) แลกกับผลบวกลวงที่เพิ่มขึ้น (specificity ลดลง) โดยค่า ROC AUC ไม่เปลี่ยนแปลง มีเพียงจุดปฏิบัติงาน (operating point) ที่เปลี่ยนไป การเลือกจุดสมดุลนี้เป็น **การตัดสินใจทางคลินิก** (ความเสียหายจากการพลาดผู้ป่วยเทียบกับภาระของการตรวจเพิ่มเติม) มิใช่ประเด็นทางการเขียนโปรแกรม ตัวเลขจากชุดทดสอบมีความแปรปรวน (ชุดทดสอบมีผู้ป่วยโรคเบาหวานประมาณ 54 ราย) จึงอาจแตกต่างจากตารางเพียงเล็กน้อย

````python
# 🏆 ส่งผลค่าอ้างอิง (baseline) — ไปป์ไลน์อ้างอิง fit กับแถวที่ไม่ใช่ชุดทดสอบเท่านั้น
from sklearn.base import clone

test_ids = pd.read_csv(lab_data("pima/test_ids.csv"))["id"]
train_mask = ~X.index.isin(test_ids)

baseline_pipe = clone(pipe)                                   # สำเนาไปป์ไลน์อ้างอิงจากขั้นที่ 3 (ยังไม่ผ่านการฝึก)
baseline_pipe.fit(X[train_mask], y[train_mask])               # ห้าม fit กับแถวของชุดทดสอบ
prob = baseline_pipe.predict_proba(X.loc[test_ids])[:, 1]

pd.DataFrame({"id": test_ids.values, "prob": prob}).to_csv("lab02_submission.csv", index=False)
print("บันทึกไฟล์ lab02_submission.csv เรียบร้อย จำนวน", len(prob), "แถว")
if NAME and JOIN_CODE:
    submit_to_leaderboard("lab02_submission.csv")
else:
    print("หากต้องการส่งผล โปรดระบุ NAME และ JOIN_CODE ในเซลล์การตั้งค่า (ไม่บังคับ)")
````
